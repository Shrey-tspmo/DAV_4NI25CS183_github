```python
# Question
data = {
    "Name": ["Asha","Bharat","Chetan","Divya","Esha"],
    "Department": ["CSE","ISE","CSE","AIML","ISE"],
    "Marks":[72,91,84,95,68],
    "Attendance": [82,94,88,96,75]
}
students = pd.DataFrame(data)
students
```

```python
#ANSS

print("\nName and Marks:")
print(students[["Name","Marks"]])

print("Students who have scored higher than 80")
higher_score = students[students["Marks"] > 80]
higher_score

print("Studnets whose attendence is higher than 85")
higher_att = students[students["Attendance"] > 80]
higher_att

print("First three students:")
display(students.iloc[:3])

print("Average marks:")
print(students["Marks"].mean())

print("Max marks:")
display(students[students["Marks"] == students["Marks"].max()])

print("Studnets from CSE department:")
display(students[students["Department"] == "CSE"])

print("Studnets who have scored above 80 and attendence is above 85%")
display(students[(students["Marks"] > 80) & (students["Attendance"] > 85)])
```
